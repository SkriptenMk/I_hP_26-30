## Funktion zur Prüfung, ob ein Jahr ein Schaltjahr ist

Die Prüfungskaskade aus der Programmierübung wird unverändert übernommen.
Anstatt das Resultat mit `print()` auszugeben, gibt die Funktion den
Wahrheitswert über `return` zurück. So kann das Resultat im aufrufenden
Code weiterverwendet werden.

```python
def check_leapyear(year: int) -> bool:
    """Prüft, ob ein Jahr nach dem gregorianischen Kalender ein Schaltjahr ist.

    Parameter:
        year: das zu prüfende Jahr

    Rückgabewert:
        True, wenn year ein Schaltjahr ist, sonst False
    """
    # Die Reihenfolge der Prüfungen ist entscheidend: Zuerst wird die
    # engste Bedingung (teilbar durch 400) geprüft, dann die Ausnahme
    # (teilbar durch 100) und erst zuletzt die Grundregel (teilbar durch 4).
    if year % 400 == 0:
        # Säkularjahre, die durch 400 teilbar sind, sind Schaltjahre (z. B. 2000).
        is_leapyear = True
    elif year % 100 == 0:
        # Übrige Säkularjahre sind keine Schaltjahre (z. B. 1900).
        is_leapyear = False
    elif year % 4 == 0:
        # Alle anderen durch 4 teilbaren Jahre sind Schaltjahre (z. B. 2024).
        is_leapyear = True
    else:
        # Alle übrigen Jahre sind keine Schaltjahre (z. B. 2001).
        is_leapyear = False

    return is_leapyear
```

```python
# Test Schaltjahr-Checker
print(check_leapyear(2001)) # Erwartetes Resultat False
print(check_leapyear(1900)) # Erwartetes Resultat False
print(check_leapyear(2000)) # Erwartetes Resultat True
```

Da die Bedingungen selbst bereits Wahrheitswerte liefern, lässt sich die
Funktion auch in einer einzigen Zeile formulieren. Die Lösung oben ist
jedoch leichter nachvollziehbar.

```python
def check_leapyear_short(year: int) -> bool:
    """Kurzform von check_leapyear mit einem einzigen logischen Ausdruck."""
    # Schaltjahr, wenn durch 4, aber nicht durch 100 teilbar –
    # oder wenn durch 400 teilbar.
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0
```

```python
# Test der Kurzform
print(check_leapyear_short(2001)) # Erwartetes Resultat False
print(check_leapyear_short(1900)) # Erwartetes Resultat False
print(check_leapyear_short(2000)) # Erwartetes Resultat True
```

## Funktion zur Berechnung von $n!$

Die `for`-Schleife aus der Programmierübung wird in die Funktion
übernommen. Der Sonderfall $0! = 1$ muss dabei nicht separat behandelt
werden: Für `n = 0` ist `range(1, 1)` leer, die Schleife wird also kein
einziges Mal durchlaufen und `f` behält den Startwert `1`.

```python
def factorial(n: int) -> int:
    """Berechnet die Fakultät n! einer natürlichen Zahl n.

    Parameter:
        n: natürliche Zahl (einschliesslich 0)

    Rückgabewert:
        das Produkt 1 · 2 · ... · n; für n = 0 der Wert 1
    """
    # Startwert 1 ist das neutrale Element der Multiplikation.
    # Gleichzeitig ist damit 0! = 1 bereits abgedeckt.
    f = 1

    # Multipliziert f nacheinander mit 1, 2, ..., n.
    # range(1, n+1) endet vor n+1, enthält also n noch.
    for i in range(1, n + 1):
        f *= i

    return f
```

```python
# Test für die Berechnung der Fakultäten
print(factorial(5)) # Erwartetes Resultat 120
print(factorial(0)) # Erwartetes Resultat 1
```

## Funktion zur Kontrolle der Promotionsbedingungen

Die Notenliste wird der Funktion als Parameter übergeben. Die Funktion
prüft die beiden Bedingungen gemäss § 9 des Promotionsreglements und gibt
`True` zurück, wenn die Promotion erfüllt ist, andernfalls `False`. Der
Type Hint `list[float]` zeigt an, dass eine Liste von
Fliesskommazahlen erwartet wird.

```python
def check_term_result(grades: list[float]) -> bool:
    """Prüft, ob die Promotionsbedingungen gemäss § 9 erfüllt sind.

    Parameter:
        grades: Liste mit den Noten aller Promotionsfächer

    Rückgabewert:
        True, wenn die Promotionsbedingungen erfüllt sind, sonst False
    """
    insufficient = 0    # Anzahl Noten unter 4
    minus_points = 0    # Summe der Abweichungen von 4 nach unten
    plus_points = 0     # Summe der Abweichungen von 4 nach oben

    # Jede Note wird einzeln mit der Grenze 4 verglichen.
    for grade in grades:
        if grade < 4:
            insufficient += 1
            minus_points += 4 - grade
        elif grade > 4:
            plus_points += grade - 4
        # Eine Note von genau 4 verändert keinen der Zähler.

    # Bedingung a: doppelte Minuspunkte nicht grösser als die Pluspunkte
    condition_a = 2 * minus_points <= plus_points
    # Bedingung b: höchstens drei ungenügende Noten
    condition_b = insufficient <= 3

    # Beide Bedingungen müssen gleichzeitig erfüllt sein.
    return condition_a and condition_b
```

```python
# Test der Promotionsbedingungen
# faecher = ['D', 'F', 'E', 'M', 'NT', 'WR', 'G', 'GG', 'MB', 'I']
print(check_term_result([3.5, 4, 5, 3, 4, 4.5, 4.5, 4.5, 5, 3.5]))   # Erwartetes Resultat False
print(check_term_result([3.5, 4, 5, 3.5, 4, 4.5, 5, 4.5, 5, 4.5]))   # Erwartetes Resultat True
print(check_term_result([3.5, 3.5, 3.5, 3.5, 6, 6, 6, 6, 6, 6]))     # Erwartetes Resultat False
```

Weil die Funktion einen Wahrheitswert zurückgibt, kann sie direkt in
einer `if`-Anweisung verwendet werden:

```python
noten = [3.5, 4, 5, 3.5, 4, 4.5, 5, 4.5, 5, 4.5]

if check_term_result(noten):
    print('Die Promotionsbedingungen sind erfüllt.')
else:
    print('Die Promotionsbedingungen sind nicht erfüllt.')
```
