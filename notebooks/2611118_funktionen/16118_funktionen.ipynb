# Funktionen in Python

Wenn man programmiert, sollten sich wiederholende Programmanweisungen
vermieden werden. Dieser Grundsatz ist so wichtig, dass er eine eigene
Abkürzung erhalten hat: **DRY** (don't repeat yourself). Um dieses
Prinzip umzusetzen, lassen sich in Python wiederkehrende Anweisungen als
Funktionen definieren und als Ganzes wiederverwenden. 

Die Syntax für die Definition einer Funktion sieht in Python
folgendermassen aus: 

Python
def name_der_funktion(parameter1, parameter2, ..., parameterN):
    anweisungen
    return rueckgabewert
Das Schlüsselwort `def` leitet die Funktionsdefinition ein, gefolgt vom
Funktionsnamen, für den dieselben Regeln gelten wie für Variablennamen.
In den Klammern stehen optional die sogenannten Parameter. Das sind
Variablen, welche beim Aufruf der Funktion die übergebenen Werte
aufnehmen. Nach dem Doppelpunkt folgen eingerückt die eigentlichen
Anweisungen sowie optional `return`, um ein Ergebnis zurückzugeben. Wird
auf return verzichtet, liefert die Funktion automatisch den Wert None
zurück, der in Python für «kein Wert» steht.

In der folgenden Code-Zelle findet sich ein einfaches Beispiel. In der
Funktion wird der Satz des Pythagoras für die Berechnung der Hypotenuse
aus den beiden Katheten als Funktion implementiert.

```python
def pythagoras(a, b):
    c = (a**2 + b**2)**(1/2)
    return c
```

```python
pythagoras(3, 4)
```

Dieses Beispiel definiert eine Funktion namens `pythagoras`, welche die
Länge der Hypotenuse eines rechtwinkligen Dreiecks berechnet. Beim
Aufruf `pythagoras(3, 4)` übernehmen die Parameter `a` und `b` die übergebenen
Argumente `3` und `4`. Anschliessend berechnet die Funktion im eingerückten
Funktionskörper die Quadratwurzel aus `a² + b²` (über die Potenz `**(1/2)`),
speichert das Resultat in der lokalen Variable `c` und gibt es über `return`
als Fliesskommazahl (`5.0`) zurück. 

```python
def pythagoras(a: float, b: float) -> float:
    c = (a**2 + b**2)**(1/2)
    return c
```

Um auf einen Blick erkennbar zu machen, welche Datentypen eine Funktion
erwartet und was sie zurückgibt, lassen sich in Python sogenannte **Type
Hints** (Typannotationen) ergänzen: 

* **`a: float, b: float`**: Der Doppelpunkt nach dem Parameternamen gibt
  den erwarteten Datentyp der Eingabewerte an – hier jeweils eine
  Fliesskommazahl (`float`). 
* **`-> float`**: Der Pfeil nach der schliessenden Klammer beschreibt
  den Datentyp des Rückgabewerts (bei einer Funktion ohne `return`
  stünde hier `-> None`). 

Python bleibt trotz dieser Angaben eine dynamisch typisierte Sprache:
Die Type Hints werden zur Laufzeit nicht erzwungen, weshalb sich auch
Ganzzahlen wie `3` und `4` problemlos übergeben lassen. Sie dienen
jedoch als wertvolle Dokumentation im Code und ermöglichen es
Entwicklungsumgebungen (IDEs), passende Autovervollständigungen
anzubieten und Unstimmigkeiten frühzeitig zu markieren.
