# Funktionen in Python: Vorgabewerte – Lösungen

Gerade bei repetitiven Berechnungen kommt es vor, dass einzelne
Parameter einer Formel meistens, aber eben nicht immer, die gleichen
Werte aufweisen.

Als Beispiel soll eine Funktion definiert werden, welche die Fläche
eines Kreises berechnet. Standardmässig soll dabei $\pi \approx 3.14$
gerechnet werden. Es soll aber möglich sein, für $\pi$ einen anderen
Wert zu verwenden.

```python
import math

def kreisflaeche(radius: float, pi: float = 3.14) -> float:
    """Berechne die Fläche eines Kreises.

    Die Fläche wird nach der Formel A = pi * r**2 berechnet. Für die
    Kreiszahl wird standardmässig der Näherungswert 3.14 verwendet.
    Bei Bedarf kann ein genauerer Wert (z. B. ``math.pi``) oder eine
    andere Näherung (z. B. ``22 / 7``) übergeben werden.

    Parameters
    ----------
    radius : float
        Der Radius des Kreises. Der Wert sollte nicht negativ sein.
    pi : float, optional
        Der Wert, der für die Kreiszahl verwendet wird. Vorgabewert
        ist 3.14.

    Returns
    -------
    float
        Die Fläche des Kreises in der Quadrateinheit des Radius.

    Examples
    --------
    >>> kreisflaeche(2)
    12.56
    >>> kreisflaeche(2, pi=3.0)
    12.0
    """
    return pi * radius ** 2
```

Dadurch, dass in der Definition der Funktion `kreisflaeche` der Variable
`pi` mit `= 3.14` der Wert 3.14 zugewiesen wird, wird standardmässig
dieser Wert verwendet. Wie alle Variablen, kann aber auch `pi`
überschrieben werden.

```python
print(kreisflaeche(2))                # 12.56            → Vorgabewert 3.14
print(kreisflaeche(2, math.pi))       # 12.566370614...  → exakter Wert
print(kreisflaeche(2, pi=22/7))       # 12.571...        → Näherungsbruch
```

In den folgenden Übungen soll das vertieft werden.

#### Übung 1: Der fehlende Winkel

*Formelsammlung S. 86 – Winkelsumme*

In jedem Dreieck gilt

$$\alpha + \beta + \gamma = 180° .$$

**Aufgabe**

1. Ergänzen Sie die Funktion `dritter_winkel(alpha, beta)`, die aus zwei
   Innenwinkeln (in Grad) den dritten berechnet. Schreiben Sie
   Typ-Hinweise und einen Docstring wie im Beispiel `kreisflaeche`.
2. Ein Dreieck hat die Winkel $\alpha = 48°$ und $\beta = 77°$. Wie gross
   ist $\gamma$?

```python
def dritter_winkel(alpha: float, beta: float) -> float:
    """Berechne den dritten Innenwinkel eines Dreiecks.

    Die Berechnung beruht auf der Winkelsumme im Dreieck
    alpha + beta + gamma = 180° (Formelsammlung S. 86).

    Parameters
    ----------
    alpha : float
        Erster Innenwinkel in Grad.
    beta : float
        Zweiter Innenwinkel in Grad.

    Returns
    -------
    float
        Der dritte Innenwinkel in Grad.

    Examples
    --------
    >>> dritter_winkel(50, 60)
    70
    """
    return 180 - alpha - beta


print(dritter_winkel(48, 77))
```

```python
# Kontrolle – diese Zelle sollte ohne Fehlermeldung durchlaufen
assert dritter_winkel(50, 60) == 70
assert dritter_winkel(90, 45) == 45
assert dritter_winkel(48, 77) == 55
print("Übung 1: alle Tests bestanden")
```

#### Übung 2: Giebelwand

*Formelsammlung S. 87 – Flächeninhalt*

Die Giebelwand eines Hauses ist ein Dreieck. Die Breite des Hauses bildet
die Grundseite $a$, die Höhe des Giebels ist $h_a$. Für die Fläche gilt

$$A = \frac{1}{2}\,a\,h_a .$$

**Aufgabe**

1. Definieren Sie eine Funktion `flaeche_dreieck(a, h_a)` mit
   Typ-Hinweisen und Docstring.
2. Die Giebelwand ist 9.6 m breit und 4.5 m hoch. Wie viele m² müssen
   gestrichen werden?

```python
def flaeche_dreieck(a: float, h_a: float) -> float:
    """Berechne die Fläche eines Dreiecks aus Grundseite und Höhe.

    Es gilt A = 1/2 * a * h_a (Formelsammlung S. 87).

    Parameters
    ----------
    a : float
        Länge der Grundseite.
    h_a : float
        Höhe auf die Grundseite a.

    Returns
    -------
    float
        Die Fläche des Dreiecks in der Quadrateinheit der Eingaben.

    Examples
    --------
    >>> flaeche_dreieck(6, 4)
    12.0
    """
    return a * h_a / 2


print(f"Fläche der Giebelwand: {flaeche_dreieck(9.6, 4.5):.2f} m²")
```

```python
# Kontrolle
assert flaeche_dreieck(6, 4) == 12
assert flaeche_dreieck(5, 3) == 7.5
assert round(flaeche_dreieck(9.6, 4.5), 2) == 21.6
print("Übung 2: alle Tests bestanden")
```

#### Übung 3: Diagonalstrebe eines Gartentors

*Formelsammlung S. 88 – Satz von Pythagoras*

Ein rechteckiges Gartentor soll mit einer diagonalen Strebe verstärkt
werden. Die Strebe bildet mit der Breite $a$ und der Höhe $b$ des Tors ein
rechtwinkliges Dreieck; sie selbst ist die Hypotenuse $c$:

$$a^2 + b^2 = c^2 \qquad\Rightarrow\qquad c = \sqrt{a^2 + b^2} .$$

*Hinweis:* Für die Quadratwurzel braucht es keine Bibliothek, denn
$\sqrt{x} = x^{0.5}$. In Python schreibt man `x ** 0.5`.

**Aufgabe**

1. Definieren Sie eine Funktion `hypotenuse(a, b)` mit Typ-Hinweisen und
   Docstring.
2. Das Tor ist 1.2 m breit und 1.6 m hoch. Wie lang muss die Strebe sein?

```python
def hypotenuse(a: float, b: float) -> float:
    """Berechne die Hypotenuse eines rechtwinkligen Dreiecks.

    Nach dem Satz von Pythagoras gilt c = (a**2 + b**2) ** 0.5
    (Formelsammlung S. 88).

    Parameters
    ----------
    a : float
        Länge der ersten Kathete.
    b : float
        Länge der zweiten Kathete.

    Returns
    -------
    float
        Länge der Hypotenuse in der Einheit der Katheten.

    Examples
    --------
    >>> hypotenuse(3, 4)
    5.0
    """
    return (a ** 2 + b ** 2) ** 0.5


print(f"Länge der Strebe: {hypotenuse(1.2, 1.6):.2f} m")
```

```python
# Kontrolle
assert hypotenuse(3, 4) == 5
assert hypotenuse(5, 12) == 13
assert round(hypotenuse(1.2, 1.6), 6) == 2
print("Übung 3: alle Tests bestanden")
```

### Teil B: Funktionen mit Vorgabewerten

In den folgenden drei Übungen werden die Funktionen aus Teil A um einen
Parameter mit Vorgabewert erweitert. Wird das Argument beim Aufruf
weggelassen, verhält sich die Funktion so, wie es im häufigsten Fall
gewünscht ist.

#### Übung 4: Winkel in Grad oder Gon

*Formelsammlung S. 86 – Winkelsumme*

In der Schweizer Vermessung werden Winkel oft in **Gon** angegeben: Ein
voller Kreis hat 400 gon, die Winkelsumme im Dreieck beträgt deshalb
200 gon statt 180°.

**Aufgabe**

1. Erweitern Sie die Funktion aus Übung 1 zu
   `dritter_winkel(alpha, beta, winkelsumme=180)`. Wird `winkelsumme`
   weggelassen, rechnet die Funktion wie bisher in Grad. Passen Sie auch
   den Docstring an.
2. Ein Vermessungsteam misst $\alpha = 62$ gon und $\beta = 85$ gon. Wie
   gross ist $\gamma$?
3. Warum ist 180 ein sinnvoller Vorgabewert?

```python
def dritter_winkel(alpha: float, beta: float, winkelsumme: float = 180) -> float:
    """Berechne den dritten Innenwinkel eines Dreiecks.

    Die Berechnung beruht auf der Winkelsumme im Dreieck
    (alpha + beta + gamma = 180° bzw. 200 gon; Formelsammlung S. 86).

    Parameters
    ----------
    alpha : float
        Erster Innenwinkel.
    beta : float
        Zweiter Innenwinkel.
    winkelsumme : float, optional
        Winkelsumme im Dreieck in der verwendeten Einheit.
        Vorgabewert ist 180 (Grad); für Gon 200 übergeben.

    Returns
    -------
    float
        Der dritte Innenwinkel in derselben Einheit wie die Eingaben.

    Examples
    --------
    >>> dritter_winkel(50, 60)
    70
    >>> dritter_winkel(50, 60, winkelsumme=200)
    90
    """
    return winkelsumme - alpha - beta


print(f"gamma = {dritter_winkel(62, 85, winkelsumme=200)} gon")

# Zu 3: Grad ist im Alltag und im Unterricht die übliche Einheit. Wer in
# Grad rechnet, muss deshalb nichts zusätzlich angeben.
```

```python
# Kontrolle
assert dritter_winkel(50, 60) == 70              # Vorgabewert: Grad
assert dritter_winkel(50, 60, winkelsumme=200) == 90
assert dritter_winkel(50, 60, 200) == 90         # auch als Positionsargument
assert dritter_winkel(62, 85, winkelsumme=200) == 53
print("Übung 4: alle Tests bestanden")
```

#### Übung 5: Vom Plan zur Wirklichkeit

*Formelsammlung S. 87 – Flächeninhalt; S. 88 – Ähnlichkeitssätze*

Auf einem Bauplan im Massstab 1 : 100 misst man die Grundseite und die
Höhe einer dreieckigen Giebelwand in **Zentimetern**. Die reale Wand ist
zum Dreieck auf dem Plan ähnlich: Alle Längen sind um den Faktor 100
grösser.

**Aufgabe**

1. Definieren Sie eine Funktion `reale_flaeche(a, h_a, massstab=100)`. Die
   Längen werden in cm auf dem Plan übergeben, das Resultat soll die reale
   Fläche in **m²** sein. Verwenden Sie Ihre Funktion `flaeche_dreieck`
   aus Übung 2. *Hinweis:* 1 m² = 10 000 cm².
2. Auf dem Plan ist die Giebelwand 9.6 cm breit und 4.5 cm hoch. Wie gross
   ist die reale Fläche? Wie gross wäre sie bei einem Plan im Massstab
   1 : 50?
3. Um welchen Faktor ändert sich die Fläche, wenn der Massstab halbiert
   wird (1 : 50 statt 1 : 100)? Begründen Sie mit den
   Ähnlichkeitssätzen.

```python
def reale_flaeche(a: float, h_a: float, massstab: float = 100) -> float:
    """Berechne die reale Fläche eines Dreiecks aus Planmassen.

    Alle Längen auf dem Plan werden mit dem Massstab vergrössert. Weil das
    reale Dreieck zum Plandreieck ähnlich ist, wächst die Fläche mit dem
    Quadrat des Massstabs (Formelsammlung S. 88).

    Parameters
    ----------
    a : float
        Grundseite auf dem Plan in cm.
    h_a : float
        Höhe auf die Grundseite auf dem Plan in cm.
    massstab : float, optional
        Nenner des Massstabs 1 : massstab. Vorgabewert ist 100.

    Returns
    -------
    float
        Die reale Fläche in m².

    Examples
    --------
    >>> reale_flaeche(6, 4)
    12.0
    """
    flaeche_plan_cm2 = flaeche_dreieck(a, h_a)
    flaeche_real_cm2 = flaeche_plan_cm2 * massstab ** 2
    return flaeche_real_cm2 / 10_000          # 1 m² = 10 000 cm²


print(f"Massstab 1:100: {reale_flaeche(9.6, 4.5):.2f} m²")
print(f"Massstab 1:50:  {reale_flaeche(9.6, 4.5, massstab=50):.2f} m²")

# Zu 3: Die Fläche wird um den Faktor (1/2)**2 = 1/4 kleiner. Bei ähnlichen
# Dreiecken wachsen alle Längen mit dem Faktor k, die Fläche mit k**2.
```

```python
# Kontrolle
assert reale_flaeche(6, 4) == 12
assert reale_flaeche(6, 4, massstab=50) == 3
assert round(reale_flaeche(9.6, 4.5), 2) == 21.6
assert round(reale_flaeche(9.6, 4.5, 50), 2) == 5.4
print("Übung 5: alle Tests bestanden")
```

#### Übung 6: Gerundete Strebenlänge

*Formelsammlung S. 88 – Satz von Pythagoras*

Für die Werkstatt genügt es, die Länge der Strebe aus Übung 3 auf
Millimeter genau anzugeben, also auf **drei Nachkommastellen** in Metern.
Manchmal wird aber eine andere Genauigkeit gewünscht.

*Hinweis:* Die eingebaute Funktion `round(zahl, stellen)` rundet auf die
gewünschte Anzahl Nachkommastellen, z. B. `round(2.71828, 2)` → `2.72`.

**Aufgabe**

1. Erweitern Sie die Funktion aus Übung 3 zu
   `hypotenuse(a, b, nachkommastellen=3)`, die das Resultat rundet.
2. Ein Tor ist 1.15 m breit und 1.8 m hoch. Geben Sie die Strebenlänge
   einmal mit dem Vorgabewert und einmal auf Zentimeter genau an.

```python
def hypotenuse(a: float, b: float, nachkommastellen: int = 3) -> float:
    """Berechne die gerundete Hypotenuse eines rechtwinkligen Dreiecks.

    Nach dem Satz von Pythagoras gilt c = (a**2 + b**2) ** 0.5
    (Formelsammlung S. 88).

    Parameters
    ----------
    a : float
        Länge der ersten Kathete.
    b : float
        Länge der zweiten Kathete.
    nachkommastellen : int, optional
        Anzahl Nachkommastellen, auf die gerundet wird.
        Vorgabewert ist 3 (bei Metern also Millimeter).

    Returns
    -------
    float
        Die gerundete Länge der Hypotenuse.

    Examples
    --------
    >>> hypotenuse(1, 1)
    1.414
    >>> hypotenuse(1, 1, nachkommastellen=1)
    1.4
    """
    return round((a ** 2 + b ** 2) ** 0.5, nachkommastellen)


print(f"Auf mm genau: {hypotenuse(1.15, 1.8)} m")
print(f"Auf cm genau: {hypotenuse(1.15, 1.8, nachkommastellen=2)} m")
```

```python
# Kontrolle
assert hypotenuse(3, 4) == 5
assert hypotenuse(1, 1) == 1.414
assert hypotenuse(1, 1, nachkommastellen=1) == 1.4
assert hypotenuse(1.15, 1.8) == 2.136
assert hypotenuse(1.15, 1.8, 2) == 2.14
print("Übung 6: alle Tests bestanden")
```
